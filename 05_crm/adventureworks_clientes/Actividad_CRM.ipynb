# Etapa 4: Evaluación y presentacion de resultados

En esta etapa evaluamos la calidad del clustering con el método de Silhoutte y se realiza un informe de resultados que incluya entre otros aspectos la descripción del problema a resolver, la descripción de los pasos seguidos en el proyecto de análisis, conclusiones del análisis de variables y utilidad y aplicacion de este modelado en el ámbito del diseño de nuevas estrategias comerciales.

```python
## Importamos las librerías necearias para evaluar la calidad de los clusteres generados con el coeficiente de Silhouette

from sklearn.metrics import silhouette_score, silhouette_samples
```

23. Escribid el código necesario para evaluar la calidad del cluster con el método de Silhoutte

```python
## Evaluamos la calidad de los clusteres generados con Silhouette

# Calculamos la puntuación del coeficiente de Silhouette para cada punto
silhouette_avg = silhouette_score(data_cluster[seleccion_columnas], data_cluster['Cluster'])
print(f"La puntuación media del coeficiente de Silhouette es: {silhouette_avg}")
```

```python
# Calculamos las puntuaciones de Silhouette para cada punto
sample_silhouette_values = silhouette_samples(data_cluster[seleccion_columnas], data_cluster['Cluster'])

# Crear un gráfico de Silhouette
fig, ax1 = plt.subplots(1, 1, figsize=(10, 6))

# El rango del eje y se basa en el número de clusters, multiplicado por el tamaño de los datos
y_lower = 10
for i in range(4):
    # Agregar las puntuaciones de Silhouette para los puntos en el cluster i
    ith_cluster_silhouette_values = sample_silhouette_values[data_cluster['Cluster'] == i]

    # Ordenar las puntuaciones de Silhouette dentro del cluster
    ith_cluster_silhouette_values.sort()

    size_cluster_i = ith_cluster_silhouette_values.shape[0]
    y_upper = y_lower + size_cluster_i

    color = plt.cm.nipy_spectral(float(i) / 4)
    ax1.fill_betweenx(np.arange(y_lower, y_upper),
                      0, ith_cluster_silhouette_values,
                      facecolor=color, edgecolor=color, alpha=0.7)

    # Etiqueta del cluster en el gráfico de Silhouette
    ax1.text(-0.05, y_lower + 0.5 * size_cluster_i, str(i))

    # Computar el nuevo y_lower para la siguiente trama
    y_lower = y_upper + 10  # 10 para el espacio en blanco entre los plots

ax1.set_title("Gráfico de Silhouette para los distintos clusters")
ax1.set_xlabel("Valores del coeficiente de Silhouette")
ax1.set_ylabel("Etiqueta del cluster")

# La línea vertical roja para la puntuación media del coeficiente de Silhouette
ax1.axvline(x=silhouette_avg, color="red", linestyle="--")

ax1.set_yticks([])  # Limpiar el eje y
ax1.set_xticks(np.arange(-0.1, 1.1, 0.2))

plt.show()
```

A continuacion se ha llevado a cabo el proceso de modelado de datos para segmentar la base de clientes en grupos que compartiesen características similares.

Esta agrupación de clientes o clustering consigue que la compañía pueda personalizar sus estrategias comerciales orientandolas a un público objetivo específico y maximizar sus resultados.

Para llevar a cabo esta segmentacion se ha empleado la técnica de aprendizaje automático no supervisado k-means.

Primeramente se ha creado un nuevo dataset seleccionando las variables por las que se desea segmentar (BickePurchase, TotalAmount, TotalChildren, Education_Num, Occupation_Num y YearlyIncome_Num).

A Continuacion se calcula el número de clusters optimo mediante el método del codo, dandole un rango de cluster de entre 2 y 10.
Obtenemos que 4 es el número optimo de clusters.

Aplicamos k-means,  se obtienen centroides representativos de cada cluster mediante el modelo de K-Means y se agrega la columna cluster al dataset creado anteriormente.

Se realiza un analisis descriptivo de los cluster, se observan sus medias y los valores de cada variable en los cluster.

Se procede a la Validación, calculando el coeficiente de Silhouette para evaluar la cohesión y separación entre los cluster.

Se identifican los siguientes segmentos:

· Cluster 0: clientes con presupuestos bajos, que prácticamente no compran bicicletas y con ingresos anuales medios-bajos. Serían unos posibles compradores eventuales, por lo que se les deberían crear campañas para incentivar compras repetitivas.

· Cluster 1: son clientes que en su mayoría compran bicicletas, tienen unos presupuestos medios-bajos y unos ingresos anuales medios-bajos. Podrían ser unos clientes con potencial por lo que se les podrían promocionar productos de la gama más económica para incentivar las ventas.  

· Cluster 2: son clientes que sí compran bicicletas y tienen unos presupuestos y unos ingresos medios anuales medios-altos. Se les podría identificar como clientes premium, para los que se podrían diseñar programas de fidelización exclusivos.

· Cluster 3: son clientes que compran bicicletas, tienen presupuestos e ingresos medios anuales medios y tienen hijos, por lo que se les podría ofrecer descuentos para productos familiares.

Tras el estudio de los clusters se puede concluir que las campañas destinadas a la venta de bicicletas y/o productos de coste elevado se deben de enfocar a los cluster 1, 2 y 3, lo que ayuda a reducir considerablemente el publico al que dirigirse y ayuda a la personalizacion de las campañas.

Las campañas de los articulos de menor valor, como piezas o accesorios pequeños, siempre se han de dirigir al publico del cluster 0, así mismo se propone el estudio pormenorizado de la sensibilidad del cluster 0 a un programa de descuentos debido a su nivel de ingresos.

Los beneficios potenciales de haber realizado este estudio son:
- Diseño de estrategias comerciales, ya que se pueden enfocar los esfuerzos hacia los segmentos de mayor valor.

- Optimización de recursos, ya que al enfocar las campañas, se reducen los costes y mejora el retorno de la inversión.
